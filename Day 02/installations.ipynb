```python
5. Identifiers

Identifiers are names given to variables, functions, classes and other objects.

# Rules for identifiers:
1. Must begin with a letter or underscore.
2. Cannot begin with a number.
3. Can contain letters, numbers and underscores.
4. Cannot contain spaces.
5. Cannot use Python keywords.
6. Identifiers are case-sensitive.

Examples:
Valid: name, age, student_name, _value
Invalid: 1name, student name, class
```

```python
name = "Vyshnavi"
age = 22
marks = 90

print(name)
print(age)
print(marks)
```

```python
x = 10
y =20
print(x,y)
```

```python
#swaping variables
x = 10 
y = 20
x,y = y,x
print(x,y)
```

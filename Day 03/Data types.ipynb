```python
#numeric
number = 100
percentage = 9.4
c = 2+3j
c2 = complex(number)
print(c2,type(c2))
print(number,type(number))
print(percentage,type(percentage))
print(c,type(c))
```

```python
#sequence
s = "vyshnavi"
lst =[10,20,30]
t = (1,2,3,4)
t2 = (100,)
print(t2,type(t2))
print(type(s),type(lst),type(t))
```

```python
#boolean
bool1 = True
bool2 = False
print(type(bool1))
print(bool1 + bool2)
```

```python
#mapping
d = {"name":"Codegnan",
     "loc":"hyd",
     "branch":4}
print(d,type(d))
```

```python
#sets
s = {1,2,3,4}
print(s,type(s))
```

```python
#empty types
lst1 = []
t1 = ()
str1 = ""
d1 = {}
s1 =set()
print(lst1,type(lst1))
print(t1,type(t1))
print(str1,type(str1))
print(d1,type(d1))
print(s1,type(s1))
print(type(str()))
```

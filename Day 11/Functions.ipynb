```python
def alphabet_count(str1):
    vowels_count = 0
    consonents_count = 0
    for ch in str1:
        if ch.isalpha():
             if ch in "aeiouAEIOU":
                 vowels_count+=1
             else:
                 consonents_count+=1
    return [vowels_count,consonents_count]       
```

```python
res = alphabet_count("Python Project")
print(res)
```

```python
#prime numbers
def prime_numbers():
    for num in range(2, 101):
        count = 0
        for i in range(1, num + 1):
            if num % i == 0:
                count += 1
        if count == 2:
            print(num, end=" ")
prime_numbers()
```

```python
#Square pattern
def square(n):
    for i in range(n):
        for j in range(n):
            print("*", end=" ")
        print()
square(4)
```

```python
#Right triangle
def right_triangle(n):
    for i in range(1, n + 1):
        for j in range(i):
            print("*", end=" ")
        print()
right_triangle(5)
```

```python
#Pyramid
def pyramid(n):
    for i in range(1, n + 1):
        print(" " * (n - i), end="")
        print("* " * i)
pyramid(5)
```

```python
#Rhombus
def rhombus(n):
    for i in range(n):
        print(" " * (n - i - 1) + "* " * n)
rhombus(5)
```

```python
#reverse string
def reverse_string(s):
    return s[::-1]

result = reverse_string("Hello")

print("Original string:", "Hello")
print("Reversed string:", result)
```

```python
#pass by object reference -immutable data types
def f(x):
    print(id(x))
    x = x+10
    print(id(x))
    return x
    
```

```python
a =100
print(id(a))
f(a)
```

```python
#pass by object reference- mutable data type

def f1(x):
    print(id(x))
    x.append(100)
    print(id(x))
```

```python
b = [10,20,30,40]
print(id(b))
f1(b)
```

```python
#recursion
def factorial(n):
    #base condition
    if n==1 or n==0:
        return 1
    #recursive function call    
    return n*factorial(n-1)    
```

```python
factorial(6)
```

```python
def n_natural_sum_even(n):
    if n==2:
        return 2
    return n+ n_natural_sum_even(n-2)  
    
```

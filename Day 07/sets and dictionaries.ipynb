```python
s = {1,2,3,5}
s1 = {}
s2 = ()
print(type(s),type(s1),type(s2))
```

```python
s1={10,20,30,40}
s2={40,50,60}
print(s1)
print(s2)
```

```python
#union
print(s1.union(s2))
```

```python
#intersection
print(s1 & s2)
```

```python
#difference
print(s1-s2)
print(s2-s1)
```

```python
#symmetric difference
print(s1 ^ s2)
```

```python
s3={26,48,18,1,4}
s4={26}
#issuperset()
print(s3.issuperset(s4))
print(s3.issubset(s4))
print(s4.issubset(s3))
```

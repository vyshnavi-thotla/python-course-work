```python
def addition(a,b):
    return a+b
```

```python
print(addition(10,20))
```

```python
addition1 = lambda x,y : x+y
```

```python
print(addition1(10,20))
```

```python
#filters
nums = [1,2,3,4,5,6,7,8]
even = filter(lambda x:x%2==0,nums)
```

```python
print(even)
print(list(even))
```

```python
#map
multiplication_by_2 = map(lambda x : x*2,nums)
print(list(multiplication_by_2))
```

```python
m = (lambda x : x*2,nums)
```

```python
print(list(m))
```

```python
#reduce
from functools import reduce
addition = reduce(lambda x,y:x+y,nums)
```

```python
print(addition)
```

```python
nums = [1,2,3,4,5]
even_odd =map(lambda x:'even' if x%2==0 else 'odd',nums)
```

```python
print(list(even_odd))
```

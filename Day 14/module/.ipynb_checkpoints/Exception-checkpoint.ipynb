```python
try:
    a = int(input())
    b = int(input())
    print(a/b)
except ZeroDivisionError as e:
    print(e)
#print("your code has same error")
except ValueError as e:
    print(e)
else:
    print("code doesn't have any error")
finally:
    print("code executed")
    
```

```python
#raise keyword
def div(x,y):
    if y==0:
        raise ZeroDivisionError()
    return x/y
#div(3,0)
try:
    div(10,0)#error code
except:
    print("division with zero")
    
```

```python
#again we are writing the data which means,previous data will be overried/erase and new data will be added
file = open("students.txt","w")
file_data = '''hello students
            how are you?'''
file.write(file_data)
```

```python
#to prevent data loss,we need to open the file in append mode - 'a'
file = open("students.txt","a")
new_data = ("welcome to python programming")
file.write(new_data)
file.close()

#reading the data from the file
file = open("students.txt","r")
print(file.read())
file.close()
```

```python
#by using "with" the file closes itself automatically
with open("students.txt","r")as file:
    for line in file:
        print(line)
```

```python
import csv
with open("sample.csv","w")as f:
    data = [['name','age'],['Ankii',20],['Siri',18]]
    writer = csv.writer(f)
    writer.writerows(data)
```

```python
def generator(x):
    yield x
    yield x+1
    yield x+2
result = generator(10)
print(result)
for i in result:
    print(i)
#print(next(result))
```

```python
lst = [10,20,30]#interable
for i in lst:
    print(i)
```

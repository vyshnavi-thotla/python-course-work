```python
#arthmetical
x = 10
y = 100
print(x+y,x-y,x*y,x/y,x%y,x//y,y**x)
```

```python
#assignment
x = 200
print(x)
x+=100  #x=x+100
print(x)
x-=200
print(x)
x*=4
print(x)
x//=2
print(x)
x%=2
print(x)
```

```python
#relational
x = 26
y = 14
print(x<y,x>y,x>=y,x<=y,x==y,x!=y)
```

```python
print((x<y) and (x==y))
print((x>y) and (x!=y))
print((x<y) or (x>=y))
print(not(x<y))
print(not(x>y))
```

```python
#bitwise
num1 = 45
num2 =4
print(num1 & num2)
print(num1 | num2)
print(num1 ^ num2)
print(~ num2)
print(num2 << 2)
print(num2 >> 1)
```

```python
#membership
l = [10,2,3,4,5]
print(100 not in l)
print(4 in l)
print(5 not in l)
```

```python
#identity
x = 5
y = 5
print(x is y)
a = [1,2,3]
b = a
b = [1,2,3]
print(a is not b)
```

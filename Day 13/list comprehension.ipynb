```python
even = [num for num in range(1,10)if num%2==0]
print(even)
```

```python
l = [34.57,89,4,78,0,456,3,6,8]
even_odd = ['even' if num%2==0 else 'add' for num in l]
print(even_odd)
```

```python
indexes = [(i,j) for i in range(4) for j in range(4)]
print(indexes)
```

```python
name = "Codegnan"
characters = [ch for ch in name if ch not in "aeiouAEIOU"]
print(characters)
```

```python
upper_case = [ch.upper() for ch in name]
print(upper_case)
```

```python
squares = [i**2 for i in range(1,11)]
print("squares:",squares)
```

```python
l = ['python','code','program']
length = [len(word)for word in l]
print("length of words:",length)
```

```python
nums = [10,15,20,30,45,50,60,75]
results = [num for num in nums if num%3==0 and num%5==0]
print("numbers divisibile by 3&5:",results)
```

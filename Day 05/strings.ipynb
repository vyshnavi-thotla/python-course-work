```python
#string
str1 = "Python"
print(str1)
print(type(str1))
```

```python
#indexing
str1 = "Python version 3.15.2"

print(str1[0])
print(str1[3])
print(str1[7])
print(str1[-1])
```

```python
#concatenation
s1 = "Python"
s2 = "Code"

s3 = s1 + s2

print(s3)
```

```python
s3 = s1 + " " + s2
print(s3)
```

```python
#repition
s1 = "Python"

print(s1 * 4)
```

```python
#methods
s = "PYTHON PROGRAMMING"

print(s.lower())
```

```python
s = "python programming"

print(s.upper())
```

```python
s = "python programming"

print(s.capitalize())
```

```python
s = "python programming language"

print(s.title())
```

```python
#checking methods
s = "PYTHON"

print(s.isupper())
print(s.islower())
print(s.isalpha())
print(s.isalnum())
```

```python
s1 = "Python"
s2 = "Python"
s3 = "Java"

print(s1 == s2)
print(s1 == s3)
print(s1 != s3)
print(s1 > s3)
```

```python
#strip
s = "   Python   "

print(s.strip())
print(s.lstrip())
print(s.rstrip())
```

```python
#split
s = "Python-DSA-is-my-favourite-subject"

result = s.split("-")

print(result)
```

```python
#join
words = ['Python', 'DSA', 'is', 'my', 'favourite', 'subject']

result = "-".join(words)

print(result)
```

```python
#built in functions
s = "Python Programming"

print(len(s))
```

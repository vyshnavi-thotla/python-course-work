```python
for i in range(4):
    for j in range(4):
        if i==0 or i==3 or j==0 or j==3:
            print("*",end = " ")
        else:
            print(" ",end =" ")
    print()    
```

```python
n = 5
for i in range(n):
    for j in range(n):
        if i==j or (i+j) ==(n-1):
           print("*",end= " ")
        else:
            print(" ",end = " ")
    print()        
```

```python
for i in range(1,5):
    for j in range(i):
        print("*",end=" ")
    print()    
    
```

```python
for i in range(1,5):
    for j in range(1,5):
        if i>=j:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()        
```

```python
num = 1
for i in range(1,5):
    for j in range(1,i+1):
        print(num,end=" ")
        num = num+1
    print()    
```

```python
n=5
for i in range(4):
    for j in range(n-i-1):
        print("*",end=" ")
    print()    
```

```python
n=5
for i in range(n):
    #spaces
    for j in range(n-i-1):
        print(" ",end=" ")
    #stars
    for j in range(2*i+1):
        print("*",end=" ")
    print()    
```

```python
n=5
for i in range(n):
    #spaces
    for j in range(i):
        print(" ",end=" ")
    #stars
    for j in range(2*n - (2*i+1)):
        print("*",end=" ")
    print()    
```

```python
#rhombus
n=5
for i in range(n):
    #spaces
    for j in range(n-i-1):
        print(" ",end=" ")
    #stars
    for j in range(2*i+1):
        print("*",end=" ")
    print()  
n=4    
for i in range(n):
    #spaces
    for j in range(i+1):
        print(" ",end=" ")
    #stars
    for j in range(2*n - (2*i+1)):
        print("*",end=" ")
    print()        
```

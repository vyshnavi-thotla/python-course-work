```python
#lists
#List is a collection of multiple values stored in a single variable.
#Lists are ordered.
#Lists are mutable (changeable).
#Duplicate values are allowed.
#Lists support different data types.
#Lists are created using square brackets [ ].
```

```python
lst = [10,'vysh',7.2,'a']
print(lst)
print(type(lst))
```

```python
#indexing
l = [10, 20, 30, 40, 50]
print(l[0])
print(l[2])
print(l[-1])
print(l[-2])
```

```python
#slicing
s = [10,20,30,40,50]
#s[0:3]
#s[:2]
#s[1:]
s[:]
```

```python
#operations
l1 = [1,2,3]
l2 = [7,8,9]
l1+l2 
```

```python
#methods
l= [1,2,3,4,5]
l[2]
```

```python
#append()
l = [1,2,4,5]
l.append(20)
print(l)
```

```python
#extend
l = [1,2,3]
l.extend([100,200,300])
print(l)
```

```python
#insert
l = [1,2,3]
l.insert(3,5)
print (l)
```

```python
#remove
l =[10, 20, 30, 40]
l.remove(40)
print(l)
#pop
l.pop(1)
print(l)
```

```python
#clear
l =[10,20,30]
l.clear()
print(l)
```

```python
#methods
mylist = [40, 10, 30, 20, 50]

mylist.sort()
print("Ascending:", mylist)

mylist.sort(reverse=True)
print("Descending:", mylist)

mylist.reverse()
print("Reverse:", mylist)

print("Count:", mylist.count(20))

print("Index:", mylist.index(30))
```

```python
#built-in functions
mylist = [10, 20, 30, 40, 50]

print("Length:", len(mylist))
print("Maximum:", max(mylist))
print("Minimum:", min(mylist))
print("Sum:", sum(mylist))
print("Average:", sum(mylist) / len(mylist))
```

```python
#tuples
#Tuples are ordered.
#Tuples are immutable (unchangeable).
#Duplicate values are allowed.
#Tuples support different data types.
#Tuples are created using parentheses ( )
t = (10, 20, 30, 40, 50)
print(t)
print(type(t))
```

```python
#indexing
t = (10, 20, 30, 40, 50)
print(t[0])
print(t[3])
print(t[-1])
```

```python
#slicing
mytuple = (10, 20, 30, 40, 50)
print(mytuple[1:4])
print(mytuple[:3])
print(mytuple[::-1])
```

```python
#built-in fuctions
t1 = (10, 20, 30, 40, 50)
print("Length:", len(t1))
print("Maximum:", max(t1))
print("Minimum:", min(t1))
print("Sum:", sum(t1))
print("Average:", sum(t1) / len(t1))
```
